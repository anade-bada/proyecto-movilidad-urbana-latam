### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).

<details>
<summary>Haz clic para ver la pista</summary>
Usa ".agg()" para aplicar funciones de promedio. Al final, reinicia el índice para mantener las columnas de la agrupación como variables (no índices).

```python

# Calcular los  promedios de trafico por ciudad, país y año

traffic_city_year_2024 = (traffic_2024
      .groupby(['city','country','year'])
      .agg({
          'jams_delay':'mean',
          'traffic_index_live':'mean',
          'jams_length_kms':'mean',
          'jams_count':'mean',
          'mins_delay':'mean',
          'travel_time_live_per_10kms_mins':'mean',
          'travel_time_hist_per_10kms_mins':'mean'
      }).reset_index()
)

# Mostrar resultado
traffic_city_year_2024.head()

```

```python
# Gráfico de barras para comparar jams_delay y city_gdp_capita por ciudad
#merged.plot( ... , y=['jams_delay', 'city_gdp_capita'])

# Agrupar por país y calcular promedios
country_data = merged_data.groupby('country').agg({
    'city_gdp_capita': 'mean',  # Promedio del GDP per cápita
    'jams_delay': 'mean'        # Promedio de jams_delay
}).reset_index()

print("Datos agrupados por país:")
print(country_data.head(15))

```

```python
# Gráfico de barras para comparar jams_delay y city_gdp_capita por ciudad
#merged.plot( ... , y=['jams_delay', 'city_gdp_capita'])
# Normalizar los datos para poder compararlos
country_data_norm = country_data.copy()
country_data_norm['gdp_norm'] = (country_data['city_gdp_capita'] / country_data['city_gdp_capita'].max()) * 100
country_data_norm['jams_norm'] = (country_data['jams_delay'] / country_data['jams_delay'].max()) * 100

# Crear gráfico de barras agrupadas
x = range(len(country_data_norm))
width = 0.35

fig, ax = plt.subplots(figsize=(12, 6))

bars1 = ax.bar([i - width/2 for i in x], country_data_norm['gdp_norm'], 
               width, label='GDP per cápita (normalizado)', color='orange', alpha=0.7)
bars2 = ax.bar([i + width/2 for i in x], country_data_norm['jams_norm'], 
               width, label='Retraso por congestión (normalizado)', color='steelblue', alpha=0.7)

ax.set_xlabel('País')
ax.set_ylabel('Índice normalizado (0-100)')
ax.set_title('Comparación GDP vs Congestión por País (Normalizado)')
ax.set_xticks(x)
ax.set_xticklabels(country_data_norm['country'], rotation=45)
ax.legend()

plt.tight_layout()
plt.show()
```

## Resumen ejecutivo 

## Contexto & objetivo:
El objetivo del análisis fue evaluar la relación entre la movilidad urbana, medida a través de indicadores de congestión y tiempos de viaje, y la productividad económica, representada por el PIB per cápita a nivel ciudad. La pregunta central busca entender si mayores niveles de congestión urbana están asociados con menores niveles de productividad económica, información clave para orientar decisiones de inversión en infraestructura de transporte y planeación urbana. Las variables clave incluyeron jams_delay, traffic_index_live, travel_time_per_10kms y city_gdp_capita, ya que permiten conectar fricción operativa diaria con resultados.

## Cobertura de datos:
El análisis se centró principalmente en 2024, integrando datos de 7 países/15 ciudades de América Latina, entre ellas Bogotá, Lima, Belo-horizonte, Ciudad de Mexico, Montevideo, Santiago, Buenos Aires,  así como otras ciudades comparables. El dataset final contiene más de 34,335 registros, con observaciones a nivel ciudad–año, incorporando indicadores de tráfico, economía, medio ambiente y contexto demográfico.

## Metodología (alto nivel):
Se realizó un proceso de limpieza y estandarización de datos, corrigiendo formatos numéricos, unificando nombres de columnas, conversión a minúsculas para consistencia, eliminación de espacos y caracteres especiales, reemplazo de acentos, estandarización de separadores (guiones bajos vs espacios), conversión de columnas numéricas desde texto a float/int,  y validando consistencia temporal. Posteriormente, los datos fueron agregados a nivel ciudad–año y se utilizó una unión INNER para integrar información de tráfico y economía, asegurando comparabilidad directa entre ambas dimensiones. Como validación, se emplearon visualizaciones exploratorias (boxplots, distribuciones y análisis de outliers) para identificar sesgos, asimetrías y eventos extremos que pudieran distorsionar promedios. ## Hallazgos iniciales:
Los resultados muestran que la relación entre congestión y productividad no es lineal. Ciudades con alta actividad económica pueden presentar congestión moderada sin afectar negativamente su PIB per cápita. Sin embargo, niveles extremos y persistentes de congestión, reflejados en grandes retrasos (jams_delay) y alta variabilidad en tiempos de viaje, se asocian con menores niveles de productividad económica. 

Se observó que los outliers de congestión (eventos severos pero poco frecuentes) concentran una parte desproporcionada del impacto negativo, lo que sugiere que el problema no es el tráfico cotidiano, sino los fallos críticos del sistema de movilidad.
